#Array Indexing

#Math Operations

```python
arr01=np.array([10,20,30,40])
arr02=np.array([2,4,6,8])
print("Addition",arr01+arr02)
print("Substraction",arr01-arr02)
print("MUltiplications",arr01*arr02)
print("Division",arr01/arr02)
```

#Numpy Functions

```python
print("Square root of arr01",np.sqrt(arr01))
print("Log of arr01",np.log(arr01))
print("sin of arr01",np.sin(arr01))
print("cos of arr01",np.cos(arr01))
print("sum of arr01",np.sum(arr01))
print("Mean value of arr01",np.mean(arr01))
print("Median value of arr01",np.median(arr01))
print("Standard deviation of arr01",np.std(arr01))
print("Variance of arr01",np.var(arr01))
print("The dot product of arr01 and arr02=",np.dot(arr01,arr02))
```
