#Heat Map


```python
import pandas as pd
import numpy as np
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create a simple 3x3 DataFrame with custom products and months
np.random.seed(42)
data = np.random.rand(3, 3) * 100

# Products as columns, Months as index
columns = ['productA', 'productB', 'productC']
months = ['Jan', 'Feb', 'March']

df = pd.DataFrame(data, columns=months, index=columns)

# Plot a simple heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(df, annot=True, cmap='BrBG', fmt='.1f', linewidths=0.5)
plt.title('Monthly Product Sales')
plt.xlabel('Months')
plt.ylabel('Products')
plt.show()
```

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
numeric_data = penguins.select_dtypes(include=['number'])
correlation_matrix = numeric_data.corr()
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='PuOr', linewidths=0.5)
```
