```python
import matplotlib.pyplot as plt
#Define the Plot
fig, axes=plt.subplots(2, 1, figsize=(7, 5),sharex=True)
#Data
months=[1,2,3,4,5,6,7,8,9]
sales=[20,24,22,28,31,30,34,28,29]
profit=[4,5,4.5,6,7,6,9,5,6]
#title
fig.suptitle("Monthly Sales and Profit")
axes[0].set_title("Sales")
axes[1].set_title("Profit")
#plot
axes[0].plot(months, sales,'r-')
axes[1].plot(months, profit, 'g-')
#labels
axes[0].set_ylabel("Sales")
axes[1].set_ylabel("Profit")
axes[1].set_xlabel("Months")
#show
plt.show()

```

#Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

#Load Tips Dataset

```python
#Load the Dataset
tips=sns.load_dataset("tips")
print(tips.head())
```

```python
tips.tail()
```

```python
tips.shape
```

#Scatter Plot

```python
sns.scatterplot(x="total_bill", y="tip", data=tips)
plt.title("Total-Bills vs Tips")
plt.xlabel("Total-Bills")
plt.ylabel("Tips")
plt.show()
```

#Line Chart

```python
sns.lineplot(x="total_bill", y="tip", data=tips)
plt.title("Total-Bills vs Tips")
plt.xlabel("Total-Bills")
plt.ylabel("Tips")
plt.show()
```

#Bar Chart

```python
sns.barplot(x="total_bill", y="tip", data=tips)
plt.title("Total-Bills vs Tips")
plt.xlabel("Total-Bills")
plt.ylabel("Tips")
plt.show()
```

#Bar Chart Horizontal

```python
sns.barplot(x="tip", y="total_bill", data=tips, orient="h")
plt.title("Total-Bills vs Tips")
plt.xlabel("Tips")
plt.ylabel("Total-Bills")
plt.show()
```
