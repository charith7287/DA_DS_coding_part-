#Imports

```python
import numpy as np
import pandas as pd
```

#Create data

```python
# Employee data
data = {
    "EmployeeID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma",
             "Frank", "Grace", "Henry", "Ivy", "Jack"],
    "Gender": ["Female", "Male", "Male", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male"],
    "Department": ["HR", "IT", "Finance", "Marketing", "Sales",
                   "IT", "HR", "Finance", "Marketing", "Sales"],
    "Age": [28, 32, 40, 30, 27, 35, 29, 45, 31, 33],
    "MonthlySalary": [35000, 50000, 65000, 42000, 38000,
                      72000, 36000, 80000, 47000, 55000]
}
```

#Create DataFrame

```python
df = pd.DataFrame(data)
```

```python
df.head()
```

#Create Yearly Salary

```python
df["Yearly_salary"] = df["MonthlySalary"] * 12
df
```

#Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
df["Gender_encoded"] = label_encoder.fit_transform(df["Gender"])
df
```

#One Hot Encoding

```python
#Create one-hot encoding on Department columns
one_hot_encoded = pd.get_dummies(df["Department"], prefix="Department")
one_hot_encoded
```

#Standardization

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df["Salary_standardized"] = scaler.fit_transform(df[["Yearly_salary"]])
df
```
