```python
#Find spaces from text
import re
text="Order ID:5678,Amount: 1299"
numbers=re.findall(r"\d+",text)
print(numbers)
```

```python
#Find spaces from text
import re
text="Order ID:5678,Amount: 1299"
numbers=re.findall(r"\s+",text)
print(len(numbers))
#
```

```python
#Find spaces from text
import re
text="Order ID:5678,Amount: 1299"
numbers=len(re.findall(r" ",text))
print(numbers)
```

#Masking with regex

```python
text01="This is my phone number:91-9876543210. My serial number is 12345"
mark_text=re.sub(r'\d{10}','**********',text01)
mark_text01=re.sub(r'\d+','X',text01)
mark_text02=re.sub(r'\d{5}','XXXXX',text01)
print(mark_text)
print(mark_text01)
print(mark_text02)
#
#
```

```python
import re

text01 = "This       is      my phone     number:91-9876543210. My serial number is 12345"
clean_text = re.sub(r'\s+', ' ', text01)
print(clean_text)
```

#Valid Pattern


```python
pattern=r'^[a-zA-Z0-9._%+-]+@[a-zA-Z0-9._]+\.[a-zA-Z]{2,}$'
email="charith@gmail.com"
if re.match(pattern,email):
  print("Valid Email")
else:
  print("Invalid Email")
```

#Python DateTime feature

```python
from datetime import datetime
current_date=datetime.now()
print("Now ==>",current_date)
```

```python
print("Year ==>",current_date.year)
print("Month ==>",current_date.month)
print("Day ==>",current_date.day)
print("Hour ==>",current_date.hour)
print("Minute ==>",current_date.minute)
```

#Formating of Date

```python
now=datetime.now()
formatted_date=now.strftime("%Y-%m-%d")
formatted_time=now.strftime("%H:%M:%S")
print(formatted_date)
print(formatted_time)
```

#Create Future Date

```python
from datetime import timedelta
print("Today's date is:",now)
no_days=int(input("Enter no of days to add:"))
future_date=now+timedelta(days=no_days)
print(f"Date after {no_days} days",future_date)
```

#Checking the number ; whether valid or invalid

```python
import pandas as pd
import re
mobile=int(input("Enter Mobile Number:"))
pattern=r'^[6-9]\d{9}$'
if re.match(pattern,str(mobile)):
  print("Valid Mobile Number")
else:
  print("Invalid Mobile Number")
#
```
