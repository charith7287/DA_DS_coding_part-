#Sort and Group Student Records

```python
#Sort the Records based on reading score and the score has to be more than 90
df[df["reading score"] >= 90].sort_values(by="reading score")
```

```python
# 1. Compare average performance in mathematics, reading, and writing
avg_scores = df[['math score', 'reading score', 'writing score']].mean()
print("=== Overall Average Performance ===")
print(avg_scores.round(2))
print("\n")

# 2. Explore performance summaries across selected categories
# Category A: Test Preparation Course
test_prep_summary = df.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean()
print("=== Performance Summary by Test Preparation Course ===")
print(test_prep_summary.round(2))
print("\n")

# Category B: Parental Level of Education
parental_edu_summary = df.groupby('parental level of education')[['math score', 'reading score', 'writing score']].mean()
print("=== Performance Summary by Parental Level of Education ===")
print(parental_edu_summary.round(2))

# 3. Export analysis results to CSV files
test_prep_summary.to_csv('performance_by_test_prep.csv')
parental_edu_summary.to_csv('performance_by_parental_education.csv')
print("\n[System Message] Results successfully exported to CSV files!")
```
