#Categorical_PLot

#Import and Create the data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.DataFrame({
    "department": ["CSE", "ECE", "CSE", "EEE", "ECE", "CSE", "EEE"],
    "score": [82, 75, 91, 68, 79, 88, 72],
    "gender": ["F", "M", "M", "F", "F", "F", "M"]
})
df
```

#Count Plot

```python
# Generate a count plot for the department column
sns.countplot(data=df, x="department",color='red')
plt.title("Department Distribution")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
```

#Bar Plot

```python
sns.barplot(data=df, x="department", y="score",color='yellow')
plt.title("Department vs. Score")
plt.xlabel("Department")
plt.ylabel("Score")
plt.show()
```

#Box Plot

```python
sns.boxplot(data=df, x="department", y="score",color='green')
plt.title("Department vs. Score")
plt.xlabel("Department")
plt.ylabel("Score")
plt.show()
#
```

#Voilin Plot

```python
sns.violinplot(data=df, x="department", y="score",inner="quartile",color='orange')
plt.title("Department vs. Score")
plt.xlabel("Department")
plt.ylabel("Score")
plt.show()
#
```
