#Creating Dataset

```python
import pandas as pd
import random
from datetime import datetime, timedelta
```

```python
# Number of synthetic records
num_records = 100
departments = ["HR", "IT", "Finance", "Marketing", "Sales", "Operations"]
cities = ["Hyderabad", "Bangalore", "Chennai", "Mumbai", "Delhi"]

data = []

for i in range(1, num_records + 1):
    employee_id = i
    name = f"Employee_{i}"
    age = random.randint(22, 60)
    department = random.choice(departments)
    salary = random.randint(30000, 150000)
    city = random.choice(cities)

    join_date = (
        datetime(2018, 1, 1) +
        timedelta(days=random.randint(0, 3000))
    ).strftime("%Y-%m-%d")

    data.append([
        employee_id,
        name,
        age,
        department,
        salary,
        city,
        join_date
    ])

# Create DataFrame
df = pd.DataFrame(
    data,
    columns=[
        "EmployeeID",
        "Name",
        "Age",
        "Department",
        "Salary",
        "City",
        "JoinDate"
    ]
)

# Save as CSV
df.to_csv("employees.csv", index=False)

print("Synthetic dataset created: employees.csv")
print(df.head())
```

#Loading Dataset

```python
# Import pandas library

import pandas as pd
print("Loading dataset...")
df = pd.read_csv("employees.csv")
print("Dataset loaded successfully.")
print("Dataset Shape:", df.shape)
```

#Correlation Calculation

```python
# Calculate correlation

print("Calculating correlation...")
correlation = df["Age"].corr(df["Salary"])
print("Correlation Value (Age vs Salary):", correlation)
print("Analysis completed successfully.")
```

#Creating a Scatter Plot

```python
# Import matplotlib

import matplotlib.pyplot as plt
print("Creating scatter plot...")
plt.scatter(df["Age"], df["Salary"])
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

#Correlation Matrix

```python
# Generate correlation matrix

print("Generating correlation matrix...")
corr_matrix = df.corr(numeric_only=True)
print(corr_matrix)
print("Matrix generated successfully.")
```

#Heatmap Visualization

```python
# Import seaborn

import seaborn as sns
import matplotlib.pyplot as plt
print("Creating heatmap...")
corr_matrix = df.corr(numeric_only=True)
sns.heatmap(corr_matrix,
            annot=True)
plt.title("Correlation Matrix")
plt.show()
print("Heatmap generated successfully.")
```

#Grouped Analysis Using Pandas

```python
# Calculate average salary by department

print("Performing grouped analysis...")
result = df.groupby("Department")["Salary"].mean()
print(result)
print("Analysis completed.")
```
