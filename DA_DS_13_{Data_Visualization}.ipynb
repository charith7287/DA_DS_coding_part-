```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from matplotlib.lines import lineStyles
#Draw a line Chart
days=[1,2,3,4,5,6,7,8,9]
temperature=[32,34,29,28,31,30,30,29,28]
plt.plot(days,temperature,color="red",marker="o",linestyle='--')
plt.title('Temperature Chart')
plt.xlabel('Days')
plt.ylabel('Temperature')
plt.show()
```

```python
# Draw a line Chart with different parameters
days = [1, 2, 3, 4, 5, 6, 7, 8, 9]
new_temperature = [20, 22, 21, 24, 23, 25, 27, 26, 24]

plt.plot(days, new_temperature, color="violet", marker="s", linestyle=":")
plt.title('Weekly Temperature Variance')
plt.xlabel('Days')
plt.ylabel('Temperature')
plt.show()
```

#Bar Chart

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#Create Data

```python
products=["Computers","Phones","Tablets","Headphones"]
sales=[130,110,170,210]
```

#Draw the bar chart

```python
plt.bar(products,sales,color="violet",width=0.3)
plt.title("Sales By Products")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
# Define different products and sales data
new_products = ["Laptops", "Smartwatches", "Monitors", "Keyboards", "Printers"]
new_sales = [240, 150, 190, 85, 120]

# Plotting the bar chart
plt.bar(new_products, new_sales, color="green", width=0.4)
plt.title("Electronics Sales Performance")
plt.xlabel("Product Categories")
plt.ylabel("Units Sold")
plt.show()
```

```python
departments=["Computer Science","Statistics","Physics","Mathematics"]
students=[120,75,60,95]
plt.barh(departments,students,color="teal")
plt.title("Students By Departments")
plt.xlabel("Departments")
plt.ylabel("Students")
plt.show()
```

#Builting  a Scatter Plot


```python
hours_studied=[1, 2, 3, 4, 5, 6, 7, 8]
marks=[45,49,55,60,64,68,75,80]
#Draw a scatter plot
plt.scatter(hours_studied,marks,color="green",marker="*")
plt.title("Marks Vs Hours Studied")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
hours = [1, 2, 2, 3, 3, 4, 4, 5, 6, 7]
marks = [35, 42, 48, 50, 57, 60, 66, 72, 78, 88]
plt.figure(figsize=(8, 5))
plt.scatter(hours, marks, color="teal", s=80, alpha=0.75,
            marker="o", edgecolors="black")
plt.title("Study Hours and Exam Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")
plt.grid(True, linestyle=":", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
hours_group_a = [1, 2, 3, 4, 5]
marks_group_a = [40, 50, 55, 65, 72]
hours_group_b = [1, 2, 3, 4, 5]
marks_group_b = [45, 48, 62, 68, 80]
plt.scatter(hours_group_a, marks_group_a, label="Group A", alpha=0.8)
plt.scatter(hours_group_b, marks_group_b, label="Group B", alpha=0.8)
plt.title("Study Hours by Group")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
