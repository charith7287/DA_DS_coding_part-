#With Open

```python
with open("myfile.txt", "w") as file:
    file.write("Hello World!\n")
    file.write("This is my first text file.")
```

#Exception Handling

```python
#Try to produce some run time error
my_list=[1,2,3,4,5]
try:
  a=int(input("Enter the number"))
  print(10/a)
  print(my_list[5])
except ZeroDivisionError:
  print("You cannot divide by a zero ")
except ValueError:
  print("Invalid input. Please enter a valid number.")
except IndexError:
    print("Index out of range")
else:
  print("The division is Successfully")
finally:
  print("This is end of the program")
```
