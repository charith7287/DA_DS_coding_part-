#Histogram

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Ensure the dataset is loaded
tips = sns.load_dataset("tips")

sns.histplot(
    data=tips,
    x='time',
    bins=10,
    color='red',
)
plt.title("Histogram of Time")
plt.xlabel("Time")
plt.ylabel("Frequency")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Ensure the dataset is loaded
tips = sns.load_dataset("tips")

# Set up the plotting area with two subplots
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Histogram for Day
sns.histplot(data=tips, x='day', ax=axes[0], color='skyblue')
axes[0].set_title('Histogram of Days')
axes[0].set_xlabel('Day')
axes[0].set_ylabel('Frequency')

# Histogram for Size
sns.histplot(data=tips, x='size', ax=axes[1], discrete=True, color='lightgreen')
axes[1].set_title('Histogram of Size')
axes[1].set_xlabel('Size')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
