#Histogram using Penguin Dataset

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
#Load the dataset
penguins = sns.load_dataset("penguins")
print(penguins.head())
```

#Dropping the Null values

```python
penguins.dropna(subset=['flipper_length_mm'])
```

```python
sns.histplot(data=penguins, x="flipper_length_mm", hue="species")
plt.show()
```

### 1. `bins`
Controls the number of intervals (bars) dividing the data range.

```python
flipper_data = penguins['flipper_length_mm'].dropna()

plt.figure(figsize=(6, 3))
plt.hist(flipper_data, bins=5)  # Try changing this to 30 to see more detail
plt.title("bins=5")
plt.show()
```

### 2. `color` & `edgecolor`
`color` sets the fill color of the bars; `edgecolor` defines the boundary outlines.

```python
plt.figure(figsize=(6, 3))
plt.hist(flipper_data, bins=15, color='lightgreen', edgecolor='darkgreen')
plt.title("color='lightgreen', edgecolor='darkgreen'")
plt.show()
```

### 3. `alpha`
Sets the opacity/transparency level between 0 (fully transparent) and 1 (fully opaque).

```python
plt.figure(figsize=(6, 3))
plt.hist(flipper_data, bins=15, alpha=0.4, color='purple')
plt.title("alpha=0.4 (Semi-transparent)")
plt.show()
```

### 4. `histtype`
Changes how the histogram bars are structured (e.g., standard `'bar'` or hollow `'step'`).

```python
plt.figure(figsize=(6, 3))
plt.hist(flipper_data, bins=15, histtype='step', edgecolor='red', linewidth=2)
plt.title("histtype='step'")
plt.show()
```

### 5. `density`
Normalizes the bins so that the area under the histogram integrates to 1, showing probability density instead of raw counts.

```python
plt.figure(figsize=(6, 3))
plt.hist(flipper_data, bins=15, density=True, color='orange')
plt.title("density=True (Probability Density instead of Counts)")
plt.show()
```

### 6. `orientation`
Sets whether the bars rise vertically (default) or grow horizontally.

```python
plt.figure(figsize=(6, 4))
plt.hist(flipper_data, bins=15, orientation='horizontal', color='teal', edgecolor='black')
plt.title("orientation='horizontal'")
plt.show()
```

### 7. `label`
Attaches a label identifier, typically displayed inside a legend.

```python
plt.figure(figsize=(6, 3))
plt.hist(flipper_data, bins=15, label='Adelie/Chinstrap/Gentoo')
plt.legend()
plt.title("Using label and plt.legend()")
plt.show()
```

### 8. `rwidth`
Specifies the relative width of the bars as a fraction of the bin size, leaving gaps between them.

```python
plt.figure(figsize=(6, 3))
plt.hist(flipper_data, bins=15, rwidth=0.7, color='coral')
plt.title("rwidth=0.7 (Gaps between bars)")
plt.show()
```

### 9. `cumulative`
Calculates a cumulative sum over the bins so that each bar represents the total value of that bin plus all preceding ones.

```python
plt.figure(figsize=(6, 3))
plt.hist(flipper_data, bins=15, cumulative=True, color='pink', edgecolor='black')
plt.title("cumulative=True (Running Total)")
plt.show()
```

```python
# Drop NaN values from the flipper_length_mm column to avoid plotting errors
flipper_data = penguins['flipper_length_mm'].dropna()

plt.figure(figsize=(10, 6))

# Creating a histogram demonstrating the requested attributes:
# bins, color, edgecolor, alpha, histtype, density, orientation, label, rwidth, cumulative
plt.hist(
    flipper_data,
    bins=15,                   # Number of intervals (bars)
    color='skyblue',           # Sets the bar color
    edgecolor='darkblue',      # Sets the border color of bars
    alpha=0.7,                 # Controls transparency (0-1)
    histtype='bar',            # Defines histogram type ('bar', 'barstacked', 'step', 'stepfilled')
    density=True,              # Displays probability density instead of frequency
    orientation='vertical',    # Sets direction ('vertical' or 'horizontal')
    label='Flipper Length',    # Adds a label to the histogram
    rwidth=0.85,               # Controls the relative width of bars (gaps between bars)
    cumulative=False           # Creates a cumulative histogram if True (set to False for standard comparison)
)

plt.title("Customized Penguin Flipper Length Histogram (Density)")
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Density")
plt.legend()
plt.show()
```

#Choose some other numeric column from the pernguin dataset and draw a Histogram

```python
# Select 'body_mass_g' as the new numeric column and drop missing values
body_mass_data = penguins['body_mass_g'].dropna()

plt.figure(figsize=(10, 6))

# Draw a customized histogram
plt.hist(
    body_mass_data,
    bins=20,
    color='plum',
    edgecolor='indigo',
    alpha=0.75,
    rwidth=0.85
)

plt.title("Distribution of Penguin Body Mass (g)")
plt.xlabel("Body Mass (g)")
plt.ylabel("Frequency")
plt.grid(axis='y', alpha=0.3)
plt.show()
```
